```python
path = "/content/drive/MyDrive/Library_Usage.csv"
import pandas as pd
df = pd.read_csv(path)
df.head()
```

```python
df.info()
```

```python
df["Patron Type Definition"].value_counts()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().sort_index().plot(
    kind="pie", #perhaps here
    figsize=(11,8),)
```

```python
plt.title("Patron Registrations by Year")
plt.xlabel("Year")
plt.ylabel("Number of Registrations")
plt.tight_layout()
plt.show()
```

```python
df["Circulation Active Month"].value_counts().plot(
    kind="bar",
    figsize=(11,8),
    color=["lime"]
)

plt.title("Library Circulation Activity by Month")
plt.ylabel("Number of Circulations")
plt.xlabel("Month")
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"]

df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    figsize=(10, 6),
    color="aquamarine",
    edgecolor="black"
)
```

```python
month_order = ["January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"]

df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    figsize=(10, 6),
    color="aquamarine",
    edgecolor="black")
plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
df['Circulation Active Month Cleaned'] = df['Circulation Active Month'].fillna('Not Captured')
df.isna().sum()
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December", "Not Captured"]
month_counts = df["Circulation Active Month Cleaned"].value_counts().reindex(month_order)

month_counts.plot(
    kind="bar",
    color="hotpink",
    edgecolor="black",
    figsize=(10,6)
)

plt.title("Library Circulation Activity by Month", fontsize=16, weight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Number of Circulations", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
#i just copied and pasted the tutorital and still managed to mess it up but mostly because I was creating new code sections and youre supposed to just add to one big one
```
